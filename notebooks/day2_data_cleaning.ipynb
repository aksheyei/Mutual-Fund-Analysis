##### IMPORTING NEW LIBRARY

```python
from sqlalchemy import create_engine
```

```python
# creating engine
engine = create_engine("sqlite:///../bluestock_mf.db")
```

```python
# CREATING dim_fund table
with engine.connect() as conn:
    conn.exec_driver_sql("""
    CREATE TABLE dim_fund(
    amfi_code INTEGER PRIMARY KEY,
    fund_house TEXT,
    scheme_name TEXT,
    category TEXT,
    sub_category TEXT,
    plan TEXT,
    launch_date TEXT,
    benchmark TEXT,
    expense_ratio_pct REAL,
    exit_load_pct REAL,
    min_sip_amount INTEGER,
    min_lumpsum_amount INTEGER,
    fund_manager TEXT,
    risk_category TEXT,
    sebi_category_code TEXT
    )""")
    conn.commit()
```

```python
# creating dim_table
with engine.connect() as conn:
    conn.exec_driver_sql("""
    CREATE TABLE dim_date(
     date_id INTEGER PRIMARY KEY,
    full_date TEXT,
    year INTEGER,
    month INTEGER,
    month_name TEXT,
    quarter INTEGER,
    day_of_week TEXT,
    is_weekend INTEGER
    )""")
    conn.commit()
```

```python
# creating fact_nav
with engine.connect() as conn:
    conn.exec_driver_sql("""
    CREATE TABLE fact_nav(
    nav_id INTEGER PRIMARY KEY,
    amfi_code INTEGER,
    date_id INTEGER,
    nav REAL,
    FOREIGN KEY (amfi_code) REFERENCES dim_fund(amfi_code),
    FOREIGN KEY (date_id) REFERENCES dim_date(date_id)
    )""")
    conn.commit()
```

```python
# creating fact_transactions
with engine.connect() as conn:
    conn.exec_driver_sql("""
    CREATE TABLE fact_transactions (
    transaction_id INTEGER PRIMARY KEY,
    investor_id TEXT,
    amfi_code INTEGER,
    date_id INTEGER,
    transaction_type TEXT,
    amount_inr INTEGER,
    state TEXT,
    city TEXT,
    city_tier TEXT,
    age_group TEXT,
    gender TEXT,
    annual_income_lakh REAL,
    payment_mode TEXT,
    kyc_status TEXT,
    FOREIGN KEY (amfi_code) REFERENCES dim_fund(amfi_code),
    FOREIGN KEY (date_id) REFERENCES dim_date(date_id)
    )""")
    conn.commit()
```

```python
# creating fact_performance table
with engine.connect() as conn:
    conn.exec_driver_sql("""
    CREATE TABLE fact_performance (
    performance_id INTEGER PRIMARY KEY,
    amfi_code INTEGER,
    return_1yr_pct REAL,
    return_3yr_pct REAL,
    return_5yr_pct REAL,
    benchmark_3yr_pct REAL,
    alpha REAL,
    beta REAL,
    sharpe_ratio REAL,
    sortino_ratio REAL,
    std_dev_ann_pct REAL,
    max_drawdown_pct REAL,
    aum_crore INTEGER,
    expense_ratio_pct REAL,
    morningstar_rating INTEGER,
    risk_grade TEXT,
    FOREIGN KEY (amfi_code) REFERENCES dim_fund(amfi_code)
    )""")
    conn.commit()
```

```python
# creating fact_aum table
with engine.connect() as conn:
    conn.exec_driver_sql("""
    CREATE TABLE fact_aum (
    aum_id INTEGER PRIMARY KEY,
    date_id INTEGER,
    fund_house TEXT,
    aum_lakh_crore REAL,
    aum_crore INTEGER,
    num_schemes INTEGER,
    FOREIGN KEY (date_id) REFERENCES dim_date(date_id)
    )""")
    conn.commit()
```

#### ANALYTICAL SQL QUERIES

####  1. Top 5 funds by AUM

```python
fact_performance.head()
```

```python
pd.read_sql(
    "SELECT d.fund_house, SUM(p.aum_crore) AS AUM "
    "FROM dim_fund d LEFT JOIN fact_performance p ON d.amfi_code = p.amfi_code "
    "GROUP BY d.fund_house "
    "ORDER BY SUM(p.aum_crore) DESC "
    "LIMIT 5;",
    engine
)
```

####  2. Average NAV per month

```python
fact_nav.head()
```

```python
dim_date.head()
```

```python
pd.read_sql(
    "SELECT d.year, d.month, AVG(n.nav) AS avg_nav "
    "FROM dim_date d LEFT JOIN fact_nav n ON d.date_id = n.date_id "
    "GROUP BY d.year, d.month "
    "ORDER BY d.year, d.month;",
    engine
)
```

#### 3: SIP Year-over-Year growth

```python
fact_transactions.head()
```

```python
pd.read_sql(
    "WITH SIP AS ( "
    "SELECT d.year as year, SUM(t.amount_inr) AS SIP_TOTAL "
    "FROM dim_date d LEFT JOIN fact_transactions t ON d.date_id = t.date_id "
    "WHERE t.transaction_type = 'SIP' "
    "GROUP BY d.year "
    "ORDER BY d.year) "
    "SELECT year, SIP_TOTAL, "
    "(SIP_TOTAL - LAG(SIP_TOTAL, 1) OVER (ORDER BY year)) * 100.0 / LAG(SIP_TOTAL, 1) OVER (ORDER BY year) AS YOY_GROWTH "
    "FROM SIP;",
    engine
)
```

#### 4. Total transactions by state

```python
pd.read_sql(
    "SELECT state,COUNT(*) AS transactions FROM fact_transactions "
    "GROUP BY state; ",engine)
```

```python
pd.read_sql(
    "SELECT f.scheme_name, f.fund_house, p.expense_ratio_pct "
    "FROM fact_performance p "
    "LEFT JOIN dim_fund f ON p.amfi_code = f.amfi_code "
    "WHERE p.expense_ratio_pct < 1 "
    "ORDER BY p.expense_ratio_pct;",
    engine
)
```

#### 6. Top 5 fund houses by total AUM

```python
pd.read_sql(
    "SELECT fund_house,SUM(aum_crore) AS AUM "
    "FROM fact_aum "
    "GROUP BY fund_house "
    "ORDER BY SUM(aum_crore) DESC "
    "LIMIT 5;",
    engine)
```

#### 7: Number of transactions by KYC status

```python
fact_transactions.head()
pd.read_sql(
    "select kyc_status,count(*) as transactions "
    "from fact_transactions "
    "group by kyc_status; ",
    engine)
```

#### 8. Average 1-year return by category

```python
pd.read_sql(
    "select f.category,avg(p.return_1yr_pct) as avg_1_year_return "
    "from dim_fund f left join fact_performance p on f.amfi_code=p.amfi_code "
    "group by f.category; ",
    engine)
```

#### 9: Funds with highest Sharpe ratio

```python
pd.read_sql(
    "SELECT f.scheme_name, f.fund_house, p.sharpe_ratio "
    "FROM fact_performance p "
    "LEFT JOIN dim_fund f ON p.amfi_code = f.amfi_code "
    "ORDER BY p.sharpe_ratio DESC "
    "LIMIT 10;",
    engine
)
```

#### 10: Monthly transaction volume trend

```python
pd.read_sql(
    "select d.year,d.month,count(t.transaction_id) as transactons "
    "from dim_date d left join fact_transactions t on d.date_id=t.date_id "
    "group by d.year,d.month "
    "order by d.year,d.month; ",
    engine)
```
